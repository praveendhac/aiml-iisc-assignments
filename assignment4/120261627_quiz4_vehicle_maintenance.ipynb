# Random Forest

```python
import pandas as pd
from scipy import stats
import numpy as np

import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
df_vdataset = pd.read_csv('vehicle_maintenance.csv')
# Q. What is the dimensionality of the dataset after loading?
print('Shape:', df_vdataset.shape)
print('Columns: ', df_vdataset.columns)
print('Total Columns: ', len(df_vdataset.columns))
print('Head:', df_vdataset.head().to_string())
print('Info: ', df_vdataset.info())
print('Describe: ', df_vdataset.describe)
```

```python
# Q. After train-test split with test size = 0.33 and random state=42, what is the dimensionality of the training dataset?

df_train, df_test = train_test_split(df_vdataset, test_size=0.33, random_state=42)
print(f"train.shape={df_train.shape}")
print(f"test.shape={df_test.shape}")
```

```python
#Q. What is the model accuracy with 4 decision trees (in percentage)? Use test size = 0.33 and random state=42

# Separate (Split) features and target
target = df_vdataset.columns[-1]
type(target)
print('Target Column/Feature:\n', target)

X = pd.get_dummies(df_vdataset.drop(columns=[target]))  # encode categorical variables
y = df_vdataset[target]
print("X Columns: ", X.columns)
print("y Series: ", y[0], "Lenght: ", len(y))

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.33,
    random_state=42
)

# Random Forest with 4 trees
rf = RandomForestClassifier(n_estimators=4, random_state=42)

# Train model
rf.fit(X_train, y_train)

# Predictions
y_pred = rf.predict(X_test)

# Accuracy
accuracy = accuracy_score(y_test, y_pred) * 100

print(f"Number of Decision Trees : 4")
print(f"Test Size                : 0.33")
print(f"Random State             : 42")
print("Model Accuracy:", accuracy, "%")
```

```python
# Q. What is the model accuracy with 100 decision trees (in percentage)? Use test size = 0.33 and random state=42
# Random Forest with 4 trees
rf_100 = RandomForestClassifier(n_estimators=100, random_state=42)

# Train model
rf_100.fit(X_train, y_train)

# Predictions
y_pred = rf_100.predict(X_test)

# Accuracy
accuracy_100 = accuracy_score(y_test, y_pred) * 100

print(f"Number of Decision Trees : 100")
print(f"Test Size                : 0.33")
print(f"Random State             : 42")
print("Model Accuracy:", accuracy_100, "%")
```
